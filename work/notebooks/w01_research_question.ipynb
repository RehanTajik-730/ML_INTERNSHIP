## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv('content_refresh_anonymized.csv')
total_rows = len(df)

ctr_opportunities = df[(df['impressions_90d'] >= 500) &
                       (df['avg_position'] > 0) &
                       (df['avg_position'] <= 20) &
                       (df['ctr'] < 0.5)]

engagement_opportunities = df[(df['sessions_90d'] >= 30) &
                              ((df['engagement_rate'] < 30) | (df['scroll_rate'] < 30))]

print(f"Total pages in starter dataset: {total_rows:,}")
print(f"High-visibility, low-CTR review candidates: {len(ctr_opportunities):,}")
print(f"Traffic-heavy, low-engagement review candidates: {len(engagement_opportunities):,}")

total_opportunities = len(ctr_opportunities) + len(engagement_opportunities)
pct_of_inventory = (total_opportunities / total_rows) * 100

print(f"\nWhy this matters for the next 7 weeks:")
print(f"These {total_opportunities:,} pages ({pct_of_inventory:.1f}% of the dataset) represent trapped value.")
print(f"A machine learning model can help us rank this exact pool so human editors tackle the most lucrative gaps first.")
```
